# Classes

```python
class Person():
    def __init__(self, name, age,color):
        self.name = name
        self.age = age
        self.color = color

    def year_of_birth(self):
        return 2025-self.age
    def project_age(self,years):
        return self.age+years
```

```python
new_person = Person('Alejandro',26,'Red')
new_person.name
```

```python
new_person.year_of_birth()
```

# Inheritance

```python
class Astronaut(Person):
    def __init__(self, name, age, color, mission_length_in_months):
        # Super Method
        super().__init__(name, age,color)
        self.mission_length_in_months = mission_length_in_months

    def age_of_return(self):
        return self.project_age(int(self.mission_length_in_months/12))
```

```python
new_astronaut = Astronaut('Alejandro',26,'Red',45)
```

```python
new_astronaut.year_of_birth()
```

```python
new_astronaut.age_of_return()
```
